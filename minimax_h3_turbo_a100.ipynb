### 💾 Salvar Vídeos no Google Drive (Opcional)

```python
# @title 4. 📂 Sincronizar com Google Drive
from google.colab import drive
import shutil, glob, os

drive.mount("/content/drive")
dest_dir = "/content/drive/MyDrive/MiniMax_Videos"
os.makedirs(dest_dir, exist_ok=True)

videos = glob.glob("/tmp/h3-outputs/*.mp4")
if not videos:
    print("Nenhum vídeo novo encontrado em /tmp/h3-outputs/")
else:
    for v in videos:
        shutil.copy(v, dest_dir)
        print(f"✅ Salvo no Drive: {os.path.basename(v)}")
    print(f"🎉 Todos os vídeos foram salvos em: {dest_dir}")
```
