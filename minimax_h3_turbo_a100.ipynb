### 📥 Downloads e Backup dos Vídeos

```python
# @title 4. 📥 Baixar Último Vídeo Gerado para o Computador
from google.colab import files
import glob, os

videos = sorted(glob.glob("/tmp/h3-outputs/*.mp4") + glob.glob("/content/*.mp4"))
seen = set()
unique = []
for v in videos:
    b = os.path.basename(v)
    if b not in seen and b.endswith(".mp4"):
        seen.add(b)
        unique.append(v)

if not unique:
    print("❌ Nenhum vídeo gerado ainda. Gere no Studio primeiro!")
else:
    latest = unique[-1]
    print(f"🎬 Disparando download: {os.path.basename(latest)}")
    files.download(latest)
```

```python
# @title 5. 📂 Sincronizar com Google Drive (Opcional)
from google.colab import drive
import shutil, glob, os

drive.mount("/content/drive")
dest_dir = "/content/drive/MyDrive/MiniMax_Videos"
os.makedirs(dest_dir, exist_ok=True)

videos = glob.glob("/tmp/h3-outputs/*.mp4")
if not videos:
    print("Nenhum vídeo novo encontrado em /tmp/h3-outputs/")
else:
    for v in videos:
        shutil.copy(v, dest_dir)
        print(f"✅ Salvo no Drive: {os.path.basename(v)}")
    print(f"🎉 Todos os vídeos foram salvos em: {dest_dir}")
```
